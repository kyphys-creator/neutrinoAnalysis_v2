```python
a_b.plot_band_comparison(
    {r'2$\sigma$ CL: Exp Bkg':    'T330/scenario_bkg_b/bands/band_*_bkgpen.json',
     r'2$\sigma$ CL: No Bkg': 'T330/scenario_bkg_a/bands/band_*_bkgpen.json',},
    level=0.954,
    optimized={'Exp Bkg': a_b},   # Overlay each scenario's optimize result
    ylim=(0, 2e12), save=True,
    fname='5eVband_comparison_T330_2sigma_bkgpen.pdf',
    logy = False,
    style = 'both',
    degeneracy_files='T330/scenario_bkg_b/degeneracy/degeneracy_bkgb_idx*.json',
    degeneracy_label=r'Degeneracy $\Delta\chi_{\text{min}}^2<10^{-3}$',
    show_errorbars=False,
    bestfit_marker= '.'
)
import matplotlib.pyplot as plt
plt.xlim(0.35, 3)
plt.ylim(1e-3, 2.5)
plt.ylabel(r"$\Phi$ [$10^{{12}}$ cm$^{{-2}}$sec$^{{-1}}$]")
plt.legend(loc='upper right', fontsize=15, frameon=False)

from matplotlib.ticker import FixedLocator, FuncFormatter

def offset_yaxis(ax, offset, step):
    """Shift the *displayed* y tick labels by `offset` (data untouched).
    Ticks are placed so that the displayed numbers are multiples of `step`."""
    lo, hi = ax.get_ylim()
    shown = np.arange(np.ceil((lo + offset) / step) * step, (hi + offset) + 1e-9, step)
    ax.yaxis.set_major_locator(FixedLocator(shown - offset))
    ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f'{v + offset:g}'))

Y_OFFSET = 457534351381.1339 / 1e12          # 足したい数（10^12 cm^-2 s^-1 単位）

offset_yaxis(plt.gca(), Y_OFFSET, step=0.5)
plt.savefig('5eVband_comparison_T330_2sigma_bkgpen.pdf')
```

```python
a_b.plot_band_comparison(
    {r'2$\sigma$ CL: Exp Bkg':    'T330/scenario_bkg_b/bands/band_*_bkgpen.json',
     r'2$\sigma$ CL: No Bkg': 'T330/scenario_bkg_a/bands/band_*_bkgpen.json',},
    level=0.954,
    optimized={'Exp Bkg': a_b},   # Overlay each scenario's optimize result
    ylim=(1e11, 8e12), save=True,
    fname='5eVband_comparison_T330_2sigma_edge_bkgpen.pdf',
    logy = True,
    style = 'both',
    degeneracy_files='T330/scenario_bkg_b/degeneracy/degeneracy_bkgb_idx*.json',
    degeneracy_label=r'Degeneracy $\Delta\chi_{\text{min}}^2<10^{-3}$',
    show_errorbars=False,
    bestfit_marker= '.'
)
plt.ylabel(r"$\Phi$ [$10^{{12}}$ cm$^{{-2}}$sec$^{{-1}}$]")
plt.legend(loc='upper right', fontsize=15, frameon=False)
from matplotlib.ticker import FixedLocator, FuncFormatter, NullFormatter, NullLocator

def offset_yaxis_log(ax, offset, shown=None):
    """Log-axis version: displayed label = value + offset (data untouched).
    Ticks are put where the displayed number is in a 1-2-5 sequence (or the
    explicit `shown` list); minor log ticks are removed so no stale labels remain."""
    lo, hi = ax.get_ylim()
    if shown is None:
        decs = np.arange(np.floor(np.log10(lo + offset)), np.ceil(np.log10(hi + offset)) + 1)
        shown = np.concatenate([np.array([1, 2, 5]) * 10.0**d for d in decs])
    shown = np.array([s for s in np.atleast_1d(shown)
                      if lo + offset <= s <= hi + offset and s - offset > 0])
    ax.yaxis.set_major_locator(FixedLocator(shown - offset))
    ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f'{v + offset:g}'))
    ax.yaxis.set_minor_locator(NullLocator())
    ax.yaxis.set_minor_formatter(NullFormatter())

Y_OFFSET = 457534351381.1339 / 1e12          # 足したい数（10^12 cm^-2 s^-1 単位）

offset_yaxis_log(plt.gca(), Y_OFFSET)
plt.savefig('5eVband_comparison_T330_2sigma_edge_bkgpen.pdf')
```
